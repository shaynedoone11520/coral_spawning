```r
# Next you want to readthe raw data and look at it
fl_coral_presence_raw <- read_csv("raw_data/CREMP_SCOR_Summaries_LTA_2024_.csv")
fl_coral_presence_raw
```

```r
# From this data I need to filter and select the data that corresponds to 2024.
    #this can be found in the tidyverse lecture
# I need to change the dates to be read as dates
fl_coral_presence_2024 <- fl_coral_presence_raw %>%
  filter(SampleYear == "2024")
fl_coral_presence_2024
dim(fl_coral_presence_2024)
```

```r
names(fl_coral_presence_2024)
# As I get further I think I may pinpoint what coral species I want to pinpoint to get more concise data set
fl_coral_cover_2024_u <- fl_coral_presence_2024 %>%
  rename(
    sample_year = SampleYear,
    sample_date = SampleDate,
    subregion = Subregion,
    habitat = Habitat,
    site_code = SiteCode,
    site_id = SiteID,
    site_name = SiteName,
    station_id = StationID
    )
fl_coral_cover_2024_u
#u stands for unfinished clean
```

```r
#need to convert dbl to date
fl_coral_cover_2024_f <- fl_coral_cover_2024_u %>%
mutate(sample_date = as.Date(sample_date, format = "%m/%d/%Y"))
# had to look up on stack over flow, make sure to include "format ="
fl_coral_cover_2024_f
```
